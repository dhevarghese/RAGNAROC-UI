```python
%%time
#Brisson
import time 
steps=int(600)
videoinput=np.zeros((27,27,1)).astype(float) 

stimulusTypes = [
    {"stimName": 1, "bu": 4*T_BU, "td": T_TD_high},
    {"stimName": 2, "bu": 0, "td": 0},
]

visualObjects = [
    {"stimulus": 1, "X":  7, "Y": 14, "latency": 0, "duration": 100}
]

maskTimes = []

for i in range(1,20):
    start_time = time.time()
    EV,LV,IG_,AM_,II,N2pc_, _= speed_check.runTrial(visualObjects, stimulusTypes, steps, videoinput, xDim = 27, yDim = 27, NNMask = i)
    end_time = time.time()
    maskTimes.append(end_time-start_time)
```

```python
#Speed Distribution Chart
import matplotlib.pyplot as plt

labels = [j for j in range(1,20)]

x = np.arange(len(labels))  
width = 0.35  

fig, ax = plt.subplots()
rects1 = ax.bar(x - width/2, maskTimes, width, label='NNMask')

ax.set_ylabel('Seconds')
ax.set_title('Walltimes by varying mask size')
ax.set_xticks(x, labels)
ax.legend()

#ax.bar_label(rects1, padding=3)

fig.tight_layout()
plt.savefig('NMask.jpeg')
plt.show()
```
